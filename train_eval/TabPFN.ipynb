# TabPFN — same wells / split / margins as `qlattice_pipeline`

Same CSV filter, `RANDOM_STATE` split, delta tuning, and train/test **dynamic-bounds** metrics as the QLattice script. **`qcr_labels`** builds label-proxy Q_cr targets inside this pipeline.

The run below writes **prediction CSVs** and **PDF/PNG Turner plots** under `notebook_exports/` (same as CLI `--plot-dir` + `--export-csv-dir`).

**Install:** `pip install tabpfn pandas numpy scikit-learn matplotlib torch` (GPU optional).

````python
%matplotlib inline

from pathlib import Path

from src.utils import default_qlattice_csv_path, run_tabpfn_from_argv

data_csv = default_qlattice_csv_path()
print(data_csv.resolve(), data_csv.is_file())

NOTEBOOK_EXPORT = Path("notebook_exports")
NOTEBOOK_EXPORT.mkdir(parents=True, exist_ok=True)
print("Exports + plots ->", NOTEBOOK_EXPORT.resolve())
````

## Run inside the notebook (same argv as CLI)

Uses `--plot-dir` and `--show-plots` so figures appear in the notebook; CSVs are written to the same folder by default.

```text
python tabpfn_standalone.py --mode qcr_labels --plot-dir notebook_exports --show-plots
```

````python
run_tabpfn_from_argv([
    "--csv", str(data_csv),
    "--mode", "qcr_labels",
    "--qcr-proxy-margin-scale", "2.5",
    "--features", "physical",
    "--tabpfn-n-estimators", "8",
    "--device", "auto",
    "--plot-dir", str(NOTEBOOK_EXPORT),
    "--export-csv-dir", str(NOTEBOOK_EXPORT),
    "--show-plots",
])
````
